**Про датасет:**

Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

**Колонки датасету:**
- `total_bill` - загальна сума рахунку (долари)
- `tip` - розмір чайових (долари)
- `sex` - стать відвідувача (Male/Female)
- `smoker` - чи курить (Yes/No)
- `day` - день тижня (Thur/Fri/Sat/Sun)
- `time` - час відвідування (Lunch/Dinner)
- `size` - розмір групи (кількість людей)

**Джерело:** Оригінальні дані зібрані в одному з американських ресторанів у 1990-х роках. Скачати і переглянути набір даних можна тут: https://www.kaggle.com/datasets/sakshisatre/tips-dataset

**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
mean_check = df.total_bill.mean().round(2)
display(f'Середній розмір рахунку cкладає: ${mean_check}')
```

```python
max_tip = df.tip.max()
display(f'Максимальні чайові були: ${max_tip}')
```

```python
is_empty = df.isna().sum()
display(is_empty)   # Порожні клітинки відсутні у даному датасеті
```

```python
lunch_ = len(df[(df['time'] == 'Lunch')])
dinner_ = len(df[(df['time'] == 'Dinner')])
together = len(df['time'].isin(['Lunch', 'Dinner']))
```

```python
display(f'За обіди: {lunch_} рахунків')
display(f'За вечері: {dinner_} рахунків')
display(f'Разом рахунків за обіди і вечері: {together}')
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
mean_bill_by_sex = df.groupby('sex')['total_bill'].mean().round(2)
mean_tips_by_day = df.groupby('day')['tip'].mean().round(2)
clients_by_day = df.groupby('day')['size'].sum()
```

```python
display(mean_bill_by_sex)   # Середній рахунок за статтю (groupby sex)
```

```python
display(mean_tips_by_day)   # Середні чайові за днем тижня (groupby day)
```

```python
display(clients_by_day)   # Відвідувачів у кожен день
```
