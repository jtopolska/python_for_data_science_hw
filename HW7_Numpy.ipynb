**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])
```

```python
matrix.shape   # форма матриці 3 рядка х 3 колонки
```

```python
matrix[0]   # перший рядок, якщо малось на увазі рядок, а не індекс 1
```

```python
matrix[:, 1]   # друга колонка, якщо малось на увазі саме колонка, а не індекс 2
```

```python
np.sum(matrix)   # сумa всіх елементів
```

```python
np.mean(matrix)   # середнє всіх елементів
```
